```python
# Make a clean copy of the production dataset
df = data3.copy()

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Remove extra spaces from crop names
df['Crop'] = df['Crop'].str.strip()

# Check the cleaned dataset
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Convert the wide dataset into a yearly dataset

years = ['2006-07', '2007-08', '2008-09', '2009-10', '2010-11']

rows = []

for _, row in df.iterrows():
    for year in years:
        rows.append({
            'Crop': row['Crop'],
            'Year': year,
            'Production': row[f'Production {year}'],
            'Area': row[f'Area {year}'],
            'Yield': row[f'Yield {year}']
        })

ml_df = pd.DataFrame(rows)

print("New dataset shape:", ml_df.shape)
print("\nFirst 10 rows:")
print(ml_df.head(10))
```

```python
# Convert Year into a numerical value
ml_df['Year'] = ml_df['Year'].str[:4].astype(int)

# Create input features (X) and target (y)
X = ml_df[['Crop', 'Year', 'Area', 'Yield']]
y = ml_df['Production']

print("Features (X):")
print(X.head())

print("\nTarget (y):")
print(y.head())

print("\nX shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Separate categorical and numerical columns
categorical_features = ['Crop']
numerical_features = ['Year', 'Area', 'Yield']

# One-hot encode Crop
preprocessor = ColumnTransformer(
    transformers=[
        ('crop', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ],
    remainder='passthrough'
)

# Transform the features
X_encoded = preprocessor.fit_transform(X)

# Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

# Create the Random Forest model
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

# Train the model
model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Make predictions
y_pred = model.predict(X_test)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Model Evaluation")
print("-----------------")
print("MAE :", mae)
print("RMSE:", rmse)
print("R² Score:", r2)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred)

plt.xlabel("Actual Production")
plt.ylabel("Predicted Production")
plt.title("Actual vs Predicted Crop Production")

# Perfect prediction line
min_value = min(y_test.min(), y_pred.min())
max_value = max(y_test.max(), y_pred.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle='--'
)

plt.show()
```

```python
# Example input
new_data = pd.DataFrame({
    'Crop': ['Rice'],
    'Year': [2011],
    'Area': [170],
    'Yield': [125]
})

# Transform the input using the same preprocessor
new_data_encoded = preprocessor.transform(new_data)

# Predict production
prediction = model.predict(new_data_encoded)

print("Predicted Production for Rice:", round(prediction[0], 2))
```

```python
# Get feature names after one-hot encoding
feature_names = preprocessor.get_feature_names_out()

# Get feature importance from Random Forest
importance = model.feature_importances_

# Create a DataFrame
feature_importance = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importance
})

# Sort by importance
feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

print(feature_importance.head(15))
```

```python
import matplotlib.pyplot as plt

# Select the top 10 important features
top_features = feature_importance.head(10)

plt.figure(figsize=(10, 6))

plt.barh(
    top_features['Feature'],
    top_features['Importance']
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Top 10 Feature Importance - Random Forest")

plt.gca().invert_yaxis()
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(n_estimators=100, random_state=42),
    "Gradient Boosting": GradientBoostingRegressor(random_state=42)
}

results = []

for name, model_test in models.items():
    model_test.fit(X_train, y_train)
    predictions = model_test.predict(X_test)

    mae = mean_absolute_error(y_test, predictions)
    rmse = np.sqrt(mean_squared_error(y_test, predictions))
    r2 = r2_score(y_test, predictions)

    results.append({
        "Model": name,
        "MAE": mae,
        "RMSE": rmse,
        "R2 Score": r2
    })

results_df = pd.DataFrame(results)

print(results_df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.bar(results_df['Model'], results_df['R2 Score'])

plt.xlabel("Machine Learning Model")
plt.ylabel("R² Score")
plt.title("Comparison of Machine Learning Models")

plt.xticks(rotation=20)
plt.ylim(0, 1)
plt.show()
```
